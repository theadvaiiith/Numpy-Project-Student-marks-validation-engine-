```python
import numpy as np

# ==================================================
# MODULE 6 — CHALLENGE 3: FINAL NUMPY PROJECT
# ==================================================


# 1. Overall student pass percentage BEFORE moderation
# Student must score >= 40 in all six subjects

passed_before_mask = np.all(marks >= 40, axis=1)

passed_before_count = np.count_nonzero(passed_before_mask)

pass_percentage_before = (
    passed_before_count / marks.shape[0]
) * 100

print("Pass percentage before moderation:", pass_percentage_before)


# 2. Overall student pass percentage AFTER moderation

passed_after_mask = np.all(updated_marks_cliped >= 40, axis=1)

passed_after_count = np.count_nonzero(passed_after_mask)

pass_percentage_after = (
    passed_after_count / updated_marks_cliped.shape[0]
) * 100

print("Pass percentage after moderation:", pass_percentage_after)


# 3. Percentage-point improvement in pass rate

pass_improvement = pass_percentage_after - pass_percentage_before

print("Pass rate improvement:", pass_improvement, "percentage points")


# 4. Top 5 students AFTER moderation

total_updated_marks = np.sum(updated_marks_cliped, axis=1)

ranking = np.argsort(total_updated_marks)[::-1]

top_5_indices = ranking[:5]

top_5_students = top_5_indices + 1

top_5_totals = total_updated_marks[top_5_indices]

print("\nTop 5 Student Numbers:", top_5_students)
print("Top 5 Total Marks:", top_5_totals)


# 5. Classroom with the highest average AFTER moderation
# 10 classrooms, 10 students, 6 subjects

classroom_marks = updated_marks_cliped.reshape(10, 10, 6)

classroom_averages = np.mean(classroom_marks, axis=(1, 2))

best_classroom = np.argmax(classroom_averages) + 1

highest_classroom_average = np.max(classroom_averages)

print("\nClassroom Averages:", classroom_averages)
print("Best Classroom:", best_classroom)
print("Highest Classroom Average:", highest_classroom_average)


# 6. Count students whose total increased by at least 20 marks

original_totals = np.sum(marks, axis=1)

moderated_totals = np.sum(updated_marks_cliped, axis=1)

marks_increase = moderated_totals - original_totals

improved_mask = marks_increase >= 20

improved_students_count = np.count_nonzero(improved_mask)

print("\nStudents who gained at least 20 marks:", improved_students_count)


# 7. Validate all moderated marks are between 0 and 100

valid_marks = np.all(
    (updated_marks_cliped >= 0) &
    (updated_marks_cliped <= 100)
)

print("\nAre all moderated marks valid?", valid_marks)


# ==================================================
# FINAL SUMMARY
# ==================================================

print("\n========== FINAL PROJECT SUMMARY ==========")

print("Total Students:", marks.shape[0])
print("Total Subjects:", marks.shape[1])

print("Pass Percentage Before:", pass_percentage_before)
print("Pass Percentage After:", pass_percentage_after)
print("Pass Rate Improvement:", pass_improvement, "percentage points")

print("Top 5 Students:", top_5_students)
print("Top 5 Scores:", top_5_totals)

print("Best Classroom:", best_classroom)
print("Students Improved by 20+ Marks:", improved_students_count)

print("Dataset Validation:", valid_marks)

print("===========================================")
```
